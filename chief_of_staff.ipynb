```python
import sys
import os

# Set up paths to the workspace and the dependencies
sys.path.append(r"d:\My Projects\Massai Live Project Course\Masai Project Jun\Jun First week Chief of Staff")
sys.path.append(r"d:\My Projects\Massai Live Project Course\Masai Project Jun\Jun First week Chief of Staff\gmail-mcp-server")

from engine import fetch_threads
from triage import triage_inbox

# Fetch live threads
print("[*] Fetching actual Gmail threads...")
try:
    threads = fetch_threads()
    
    if not threads:
        print("[-] No threads returned from Gmail.")
    else:
        print(f"\n[+] Successfully fetched {len(threads)} threads. Triaging using Gemini...")
        results = triage_inbox(threads)
        
        print("\n" + "=" * 80)
        print("                        LIVE GMAIL TRIAGE RESULTS                       ")
        print("=" * 80)
        for i, t in enumerate(results, 1):
            priority = t.get("priority", "unknown")
            category = t.get("category", "other")
            reason = t.get("reason", "")
            
            print(f"\n#{i} [{priority.upper()}] | Category: {category.upper()}")
            print(f"  From    : {t.get('sender')}")
            print(f"  Subject : {t.get('subject')}")
            print(f"  Reason  : {reason}")
            print(f"  Snippet : {t.get('snippet')[:100]}...")
except Exception as e:
    print(f"\n[!] Error: {e}")
```
